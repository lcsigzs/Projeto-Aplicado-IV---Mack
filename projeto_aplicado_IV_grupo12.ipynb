# **Referencial Teórico**



**1.   Séries temporais**<br>
Uma série temporal é formada por observações de determinada variável realizadas periodicamente e organizadas de acordo com a ordem temporal. Diferentemente de dados sem ordenação temporal, as observações de uma série podem apresentar relação com valores observados anteriormente. O estudo de séries temporais pode ter como objetivos a investigação do comportamento dos dados, sua modelagem e a previsão de valores futuros (MORETTIN; TOLOI, 2018).<br>
A taxa de desemprego é uma variável que pode ser analisada dessa forma, pois seus valores são observados ao longo de períodos sucessivos. O comportamento da série pode apresentar diferentes características, como tendência, sazonalidade, ciclos e variações irregulares. A tendência representa a direção predominante dos valores ao longo do tempo, enquanto a sazonalidade corresponde a padrões que se repetem em intervalos regulares. Os ciclos representam movimentos de crescimento e redução que não necessariamente apresentam período fixo, enquanto o ruído corresponde às variações não explicadas pelos demais componentes (MORETTIN; TOLOI, 2018).

**2.   Autocorrelação e estacionariedade**<br>
A análise de uma série temporal também envolve a investigação da relação entre observações em diferentes momentos. A autocorrelação permite verificar a relação entre os valores atuais e valores anteriores da própria série, considerando diferentes defasagens. A função de autocorrelação (ACF) e a função de autocorrelação parcial (PACF) são ferramentas utilizadas para investigar esse comportamento e podem auxiliar na identificação de características relevantes para a modelagem.<br>
Outro aspecto importante é a estacionariedade. Uma série estacionária apresenta propriedades estatísticas que não dependem do momento em que a observação é realizada. A presença de tendência ou sazonalidade pode tornar a série não estacionária, sendo necessário investigar essa característica antes da aplicação de determinadas técnicas de modelagem (MORETTIN; TOLOI, 2018).

**3.   Técnicas utilizadas em trabalhos relacionados**<br>
Estudos sobre previsão da taxa de desemprego mostram a utilização de diferentes técnicas de séries temporais. Mahmudah (2017), por exemplo, utilizou dados anuais da taxa de desemprego da Indonésia entre 1986 e 2015 e aplicou modelos ARIMA. A autora identificou a não estacionariedade da série e utilizou diferenciação antes da modelagem, obtendo como melhor modelo o ARIMA (0,2,1). O estudo demonstra a possibilidade de utilização de modelos ARIMA em séries de desemprego, inclusive quando é necessário realizar diferenciação para tratar a não estacionariedade.<br>
Yamacli e Yamacli (2023) compararam modelos ARIMA e redes neurais artificiais para estimar a taxa de desemprego da Turquia utilizando 176 observações mensais entre 2008 e 2022. O estudo considerou também o período da pandemia de COVID-19 e mostrou que o desempenho das abordagens pode variar de acordo com o período analisado. Os resultados indicam que modelos estatísticos e métodos de aprendizado de máquina podem apresentar comportamentos diferentes diante de alterações relevantes na série.<br>
Outro trabalho relacionado é a tese de Alves (2024), que analisou as taxas de desemprego dos países do BRICS e aplicou diferentes métodos de previsão, incluindo redes neurais artificiais, ARIMA, suavização exponencial e decomposição STL. A utilização de diferentes métodos reforça a importância de considerar as características das séries antes da escolha da técnica de previsão.<br>
Os trabalhos analisados apresentam diferentes alternativas para a previsão de taxas de desemprego. Modelos como ARIMA apresentam como vantagem a utilização da própria estrutura temporal da série e podem ser adequados quando existem relações temporais bem definidas. Como limitação, seu desempenho pode ser afetado por alterações estruturais ou comportamentos que não sejam bem representados pela estrutura do modelo. Métodos de aprendizado de máquina, como redes neurais, permitem representar relações mais complexas, mas podem demandar maior quantidade de dados e uma etapa de preparação mais cuidadosa. Métodos de suavização e decomposição também podem ser utilizados quando a série apresenta componentes de tendência e sazonalidade relevantes.<br>
Dessa forma, a escolha da técnica de previsão não será definida exclusivamente a partir de um modelo específico. A análise exploratória da série será utilizada para identificar suas características e orientar a seleção das técnicas que serão implementadas nas etapas posteriores do projeto.

# **Diagrama de Solução**

O pipeline proposto organiza as principais etapas do desenvolvimento do projeto, desde a seleção e preparação dos dados até a avaliação e interpretação das previsões.<br>

Na etapa de seleção do recorte, serão consideradas as observações referentes à população de 15 anos ou mais, com sexo Total. A escolaridade será a principal dimensão de análise, considerando inicialmente as categorias Less than basic, Basic e Intermediate. A categoria Advanced não será incluída na modelagem inicial devido à presença de valores ausentes e registros classificados como Unreliable em parte do período analisado, comprometendo a continuidade da série. A etapa de verificação da qualidade dos dados tem como objetivo identificar valores ausentes, registros classificados como Unreliable, possíveis quebras de série e a continuidade das observações. Essa verificação será utilizada para avaliar a adequação das séries selecionadas aos procedimentos posteriores de análise e previsão.<br>
Em seguida, os dados serão organizados cronologicamente, garantindo que cada série apresente sua sequência trimestral de observações. A análise exploratória será utilizada para investigar o comportamento das séries, incluindo sua evolução ao longo do tempo e possíveis diferenças entre os níveis de escolaridade.<br>
Na etapa de análise das séries temporais serão investigadas características como tendência, sazonalidade, autocorrelação e estacionariedade. Poderão ser utilizadas ferramentas como ACF e PACF, além de métodos de decomposição, de acordo com as características observadas.<br>
A etapa de preparação das séries será definida a partir dos resultados da análise anterior. Caso seja necessário, poderão ser aplicadas transformações, diferenciações ou outros procedimentos para adequar os dados às técnicas selecionadas. A divisão entre dados utilizados no treinamento e na avaliação também será definida nessa etapa.<br>
Na etapa de modelagem, serão utilizados inicialmente modelos de referência, incluindo um modelo ingênuo sazonal e um modelo candidato da família SARIMA. Também serão investigadas abordagens de suavização exponencial, como ETS, e métodos de aprendizado de máquina. A definição das configurações e dos modelos que serão efetivamente utilizados será orientada pelas características identificadas na análise exploratória e pela adequação das técnicas às séries disponíveis.<br>
Após a implementação, os modelos serão avaliados e comparados utilizando dados reservados para avaliação e métricas de desempenho adequadas ao problema. Poderão ser utilizadas estratégias de validação temporal para verificar a estabilidade dos resultados. A comparação dos desempenhos será utilizada para definir o modelo empregado nas previsões finais.<br>
Por fim, serão produzidas previsões para períodos futuros e os resultados serão interpretados considerando o comportamento da taxa de desemprego e as diferenças observadas entre os níveis de escolaridade.
