# **Introdução**

O desemprego é um dos indicadores utilizados para acompanhar o mercado de trabalho e sua evolução ao longo do tempo. Esse comportamento pode variar entre diferentes grupos da população, considerando características como escolaridade, sexo e faixa etária. Analisar a taxa de desemprego considerando essas características permite observar diferenças entre os grupos e como elas se modificam ao longo dos períodos analisados.<br>
Neste projeto, será estudada a taxa de desemprego no Brasil a partir de dados trimestrais disponibilizados pelo ILOSTAT. A base escolhida corresponde ao conjunto *Unemployment rate by sex, age and education (%)* e apresenta informações sobre a taxa de desemprego de acordo com sexo, faixa etária e nível de escolaridade.<br>
O período disponível na base utilizada pelo grupo compreende o primeiro trimestre de 2012 até o primeiro trimestre de 2026. A taxa de desemprego será utilizada como variável principal, tendo os níveis de escolaridade como principal dimensão de análise. As informações de sexo e faixa etária serão utilizadas como dimensões complementares, permitindo comparar o comportamento da taxa entre diferentes grupos.<br>
A escolha do tema está relacionada à importância de acompanhar as mudanças no mercado de trabalho e às diferenças que podem ser observadas entre grupos com diferentes características. O projeto também está relacionado ao ODS 8 (Trabalho decente e crescimento econômico), definido entre os objetivos que podem orientar os projetos da disciplina.<br>
Como os dados estão organizados cronologicamente e possuem frequência trimestral, eles podem ser tratados como uma série temporal. Dessa forma, além de analisar o comportamento histórico da taxa de desemprego, será possível investigar padrões presentes nos dados e posteriormente aplicar técnicas de previsão.<br>
O problema que orienta o projeto é: como a taxa de desemprego no Brasil se comporta ao longo do tempo para diferentes níveis de escolaridade e em que medida esse comportamento pode ser utilizado para realizar previsões futuras?<br>
O objetivo geral é analisar o comportamento da taxa de desemprego no Brasil ao longo do tempo, considerando diferentes níveis de escolaridade como principal dimensão de análise e utilizando sexo e faixa etária como dimensões complementares. A partir dessa análise, pretende-se desenvolver e avaliar modelos de séries temporais capazes de realizar previsões da taxa de desemprego.<br>
Para alcançar esse objetivo, foram definidos os seguintes objetivos específicos:
*   analisar a evolução da taxa de desemprego no período disponível na base;
*   comparar o comportamento da taxa de desemprego entre diferentes níveis de escolaridade;
*   observar diferenças entre sexo e faixa etária;
*   identificar padrões relevantes nas séries temporais analisadas;
*   desenvolver modelos de séries temporais para realizar previsões;
*   comparar o desempenho dos modelos utilizados.
